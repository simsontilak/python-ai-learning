```python
students = []
while True:
    name = input("Enter Student name (or 'quit' to stop): ")
    if name.lower() == "quit":
        break

    scores = []
    for i in range(3):
        score = float(input(f"Enter score {i + 1}: "))
        scores.append(score)
    avg_score = sum(scores) / len(scores)
    students.append({"name":name, "scores": tuple(scores), "average": avg_score})
print("\nAI Student Summary:")
for student in students:
    print(f"{student['name']} - Avg Score: {student['average']:5.2f}")
        
```

```python
import random
target = random.randint(1, 10)
attempts = 0
print("Welcome to AI Guessing Game!")
print("I'm thinking of a number between 1 and 10.")

while True:
    guess = int(input("Take a guess"))
    attempts += 1

    if guess < target:
        print("Too low!")
    elif guess > target:
        print("Too hight!")
    else:
        print(f"Correct! you guessed it in {attempts} tries")
        break
```
