```python
import pandas as pd
data = pd.DataFrame({
    "Name":["Alice","Bob","Simson","Hari"],
    "Age":[25,30,35,None],
    "Score":[88,92,85,90]
})
print("Dataset Overview")
print(data)

print("\nBasic Info:")
print(data.info())

print("\nSummary Stats:")
print(data.describe())
```
